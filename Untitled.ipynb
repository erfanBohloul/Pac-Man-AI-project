```python
from datetime import datetime
from pathlib import Path
import json
import os
import random
import sys

# Run Pygame on an off-screen SDL surface so GIF capture does not open a desktop window.
os.environ['SDL_VIDEODRIVER'] = 'dummy'
os.environ['SDL_AUDIODRIVER'] = 'dummy'

import matplotlib.pyplot as plt
import numpy as np
import pygame
from PIL import Image as PILImage
from IPython.display import Image, display

from src.controller import BasicController, DQNAgent, DQNConfig, MCTSAgent
from src.environment import GhostsPacmanEnvironment
from src.evaluation import evaluate_algorithm
from src.drawer import PygameDrawer
from src.state import ActionSpaceEnum
```

```python
PROJECT = Path.cwd()

GRID_SIZE = 15
NUM_GHOSTS = 4
MAX_STEPS = 500
STABILITY_RATE = 0.75
SEED = 0
EVAL_SEED = 1_000_000
TEST_SEED = 2_000_000      # Held-out from seeds used to select the best DQN checkpoint.
EVAL_EPISODES = 10          # Increase to 100+ for a more stable final comparison.
EVAL_EVERY = 50
TRAIN_EPISODES = 1000
MCTS_ITERATIONS = 25
MCTS_DEPTH = 20
```

```python
run_id = datetime.now().strftime('%Y%m%d-%H%M%S')
OUTPUT = PROJECT / 'outputs' / run_id
OUTPUT.mkdir(parents=True, exist_ok=False)
```

```python
env = GhostsPacmanEnvironment(
    grid_size=GRID_SIZE, num_ghosts=NUM_GHOSTS, max_steps=MAX_STEPS,
    stability_rate=STABILITY_RATE, full_hash=True,
)
print(f'{NUM_GHOSTS} ghosts, {MAX_STEPS} max steps | output: {OUTPUT}')
```

```python
def record_game_gif(controller, filename, seed=20_000):
    random.seed(seed)
    frames = []
    observation = env.reset()
    action = ActionSpaceEnum.RIGHT
    drawer = PygameDrawer(grid_size=env.grid_size, cell_size=28, framerate=0, sound=False)
    try:
        while not observation.done:
            pygame.event.pump()
            action = controller.get_action(observation)
            drawer.draw(observation.map, action)
            frame = pygame.surfarray.array3d(drawer.screen).transpose(1, 0, 2)
            frames.append(PILImage.fromarray(frame))
            observation = env.step(action)
        drawer.draw(observation.map, action)
        frame = pygame.surfarray.array3d(drawer.screen).transpose(1, 0, 2)
        frames.append(PILImage.fromarray(frame))
    finally:
        drawer.close()

    gif_path = OUTPUT / filename
    frames[0].save(gif_path, save_all=True, append_images=frames[1:], loop=0, duration=200)
    print(f'{filename}: score={observation.score}, steps={observation.step_count}, '
          f'pellets_left={len(observation.map.pellets)}')
    display(Image(filename=str(gif_path)))
    return gif_path
```

## DQN

```python
dqn_config = DQNConfig(
    seed=SEED,
    navigation_features=True,
    wall_features=False,
    local_wall_features=True,
    epsilon_decay_steps=50_000,
    learning_starts=1_000,
)
dqn = DQNAgent(env, dqn_config)
best_checkpoint = dqn.train(
    env,
    episodes=TRAIN_EPISODES,
    output_dir=OUTPUT,
    eval_every=EVAL_EVERY,
    eval_episodes=EVAL_EPISODES,
    eval_seed=EVAL_SEED,
)
print(f'Best DQN checkpoint: {best_checkpoint}')
```

## DQN results

```python
rows = [json.loads(line) for line in (OUTPUT / 'training.jsonl').read_text().splitlines()]
train_episodes = np.array([row['episode'] for row in rows])
train_scores = np.array([row['score'] for row in rows])
validation = [row for row in rows if 'validation' in row]
val_episodes = [row['episode'] for row in validation]
val_scores = [row['validation']['Mean-score'] for row in validation]
val_win_rates = [row['validation']['Win-rate'] for row in validation]
rolling_scores = np.array([
    train_scores[max(0, i - 49):i + 1].mean() for i in range(len(train_scores))
])

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
axes[0].plot(train_episodes, train_scores, alpha=0.25, label='Episode score')
axes[0].plot(train_episodes, rolling_scores, label='50-episode mean')
axes[0].set(title='DQN training', xlabel='Episode', ylabel='Score')
axes[0].legend()
axes[1].plot(val_episodes, val_scores, marker='o')
axes[1].set(title=f'Greedy validation score ({EVAL_EPISODES} episodes)', xlabel='Training episode', ylabel='Mean score')
axes[1].grid(alpha=0.25)
axes[2].plot(val_episodes, val_win_rates, marker='o')
axes[2].set(title='Validation win rate', xlabel='Training episode', ylabel='Win rate')
axes[2].set_ylim(0, 1)
axes[2].grid(alpha=0.25)
fig.tight_layout()
plt.show()

[(row['episode'], round(row['validation']['Mean-score'], 1), row['validation']['Win-rate'])
 for row in validation]
```

## Compare all three agents

```python
comparison_episodes = EVAL_EPISODES
comparison_seed = TEST_SEED

dqn_eval = DQNAgent.from_checkpoint(best_checkpoint, env)
vanilla_eval = MCTSAgent(
    env,
    rollout_controller=BasicController(legal_only=True, seed=SEED + 107),
    iterations=MCTS_ITERATIONS,
    exploration=1.4,
    max_depth=MCTS_DEPTH,
    gamma=0.98,
    seed=SEED + 100,
)
dqn_rollout_eval = DQNAgent.from_checkpoint(best_checkpoint, env)
dqn_mcts_eval = MCTSAgent(
    env,
    rollout_controller=dqn_rollout_eval,
    iterations=MCTS_ITERATIONS,
    exploration=1.4,
    max_depth=MCTS_DEPTH,
    gamma=0.98,
    seed=SEED + 200,
)

agents = {
    'DQN': dqn_eval,
    'Vanilla MCTS': vanilla_eval,
    'DQN + MCTS': dqn_mcts_eval,
}
comparison = {
    name: evaluate_algorithm(env, agent, num_episodes=comparison_episodes, seed=comparison_seed)
    for name, agent in agents.items()
}

for name, metrics in comparison.items():
    print(
        f"{name}: mean score={metrics['Mean-score']:.1f} ± {metrics['Std-score']:.1f}, "
        f"food={metrics['Mean-food-fraction']:.1%}, "
        f"win={metrics['Win-rate']:.1%}, death={metrics['Death-rate']:.1%}"
    )
```

```python
labels = list(comparison)
fig, axes = plt.subplots(1, 4, figsize=(17, 4))
plot_specs = [
    ('Mean-score', 'Mean score'),
    ('Mean-food-fraction', 'Food collected'),
    ('Win-rate', 'Win rate'),
    ('Death-rate', 'Death rate'),
]
for ax, (metric, title) in zip(axes, plot_specs):
    values = [comparison[name][metric] for name in labels]
    errors = ([comparison[name]['Std-score'] for name in labels]
              if metric == 'Mean-score' else None)
    ax.bar(labels, values, yerr=errors, capsize=4,
           color=['#3973ac', '#e28e2c', '#4c9f70'])
    ax.set_title(title)
    ax.tick_params(axis='x', rotation=20)
    ax.grid(axis='y', alpha=0.25)
fig.suptitle(
    f'{comparison_episodes} held-out episodes, seeds '
    f'{comparison_seed}–{comparison_seed + comparison_episodes - 1}; '
    f'{GRID_SIZE}×{GRID_SIZE}, {NUM_GHOSTS} ghosts, {MAX_STEPS} steps; '
    f'MCTS budget {MCTS_ITERATIONS} × {MCTS_DEPTH}'
)
fig.tight_layout()
plt.show()
```
